```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import ticker
from matplotlib import cm

def pie_plot ( tmp_x: np.array, tmp_y: np.array , title: str ):
    """
    Pie ploting. In params are numpy array with values, numpy array with legend and title of graph.
    Plot pie graph with legend according to y array with differente colors.
    """
    if len(tmp_x) != len(tmp_y):
        raise Exception ('Invalid labeling.')
    plt.figure(figsize=(8, 8))
    def func(pct, allvals):
        absolute = float((pct/100.*np.sum(allvals)))
        return f"{absolute:.2f}"
    cmap = cm.get_cmap("Spectral")
    colors = cmap(np.linspace(0, 1, len(tmp_x)))
    plt.pie(tmp_x, labels=tmp_y, autopct=lambda pct:
            func(pct, tmp_x), startangle=140, colors=colors)
    plt.title(title)
    plt.show()

def bar_plot ( tmp_x: np.array, tmp_y: np.array , title: str, xlabel: str, ylabel: str ):
    """
    Bar ploting. In params are numpy arrays with values and title of graph.
    """
    fig = plt.figure(figsize=(10,8),constrained_layout=True)
    ax_ = fig.add_subplot()
    ax_.yaxis.set_minor_locator(ticker.AutoMinorLocator())
    ax_.grid(axis="y", color="black", alpha=.5, linewidth=.5)
    ax_.grid(axis="y", which='minor', color="black", alpha=.1, linewidth=.5)
    ax_.set_title(title)
    ax_.set_xticks(range(len(tmp_x)))
    ax_.set_xticklabels(tmp_x, rotation=45, ha='right')
    ax_.set_xlabel(xlabel)
    ax_.set_ylabel(ylabel)
    ax_.bar(tmp_x,tmp_y)
    plt.show()

def scatter_plot ( tmp_x: np.array, tmp_y: np.array, title: str,
                 label_x: str, label_y: str):
    """
    Scatter ploting with formating size and colors.
    Given params are tltle and axis labels.
    """
    cmap = cm.get_cmap("coolwarm")
    size = (tmp_x + tmp_y)**2 / 2
    colors = cmap(np.linspace(0, 1, len(tmp_x)))
    fig = plt.figure(figsize=(10,8),constrained_layout=True)
    ax_ = fig.add_subplot()
    ax_.scatter (tmp_x, tmp_y, c=colors, s = size)
    ax_.set_title(title)
    ax_.xaxis.set_major_locator(ticker.AutoLocator())
    ax_.grid(axis="x", color="black", alpha=.5, linewidth=.5)
    ax_.yaxis.set_major_locator(ticker.AutoLocator())
    ax_.grid(axis="y", color="black", alpha=.5, linewidth=.5)
    ax_.set_xlabel(label_x)
    ax_.set_ylabel(label_y)
    plt.show()

def box_plot ( tmp_x: np.array , title: str ):
    """
    Bor plotting with better readable format.
    Given paramater is title of chart.
    """
    fig = plt.figure(figsize=(10,8),constrained_layout=True)
    ax_ = fig.add_subplot()
    ax_.boxplot (tmp_x, patch_artist = True, boxprops = dict(facecolor = "y"),
                capprops = dict(color = "r",),
                 medianprops = dict(color = "r"))
    ax_.set_title(title)
    ax_.yaxis.set_minor_locator(ticker.AutoMinorLocator())
    ax_.grid(axis="y", color="black", alpha=.5, linewidth=.5)
    ax_.grid(axis="y", which='minor', color="black", alpha=.1, linewidth=.5)
    plt.show()

def double_plot ( tmp_x: np.array, tmp_y: np.array, legend: list,
                 list_: list, tick_labels: list ):
    """
    Two lines plotter with params of legend, title, tick_labels, axis
    labels. Also us of dot connected lines.
    """
    plt.figure(figsize=(10,6))
    plt.plot(tmp_x,'or-')
    plt.plot(tmp_y,'ob-')
    plt.xlabel(list_[0])
    plt.ylabel(list_[1])
    plt.xticks(ticks=range(len(tick_labels)), labels=tick_labels ,rotation=list_[2])
    #plt.gca().xaxis.set_minor_locator(ticker.AutoMinorLocator())
    plt.gca().yaxis.set_minor_locator(ticker.AutoMinorLocator())
    plt.grid(axis="y", color="black", alpha=.5, linewidth=.5)
    plt.grid(axis="y", which='minor', color="black", alpha=.1, linewidth=.5)
    #plt.grid(axis="x", color="black", alpha=.5, linewidth=.5)
    #plt.grid(axis="x", which='minor', color="black", alpha=.1, linewidth=.5)
    plt.legend(legend)
    plt.title(list_[3])
    plt.show()
    
def multi_plot(x_values: np.ndarray, 
               y_values: list, 
               legends: list, 
               axis_config: list, 
               tick_labels=None):
    """
    Multi-line plotter with support for:
    - x_values: common x-axis data (np.array)
    - y_values: list of np.arrays to plot
    - legends: list of legend labels for each line
    - axis_config: [xlabel, ylabel, xtick_rotation, title]
    - tick_labels: x-tick labels (strings)
    """

    plt.figure(figsize=(10, 6))

    # Plot each line with a different color and marker
    for i, y in enumerate(y_values):
        plt.plot(x_values, y, marker='o', label=legends[i])

    # Axes labels and title
    plt.xlabel(axis_config[0])
    plt.ylabel(axis_config[1])
    plt.title(axis_config[3])

    # X-axis tick labels
    plt.xticks(ticks=x_values, labels=tick_labels ,rotation=axis_config[2])

    # Minor ticks and grid
    plt.gca().yaxis.set_minor_locator(ticker.AutoMinorLocator())
    plt.grid(axis="y", color="black", alpha=.5, linewidth=.5)
    plt.grid(axis="y", which='minor', color="black", alpha=.1, linewidth=.5)

    plt.legend()
    plt.tight_layout()
    #plt.savefig(axis_config[3]+".jpg")
    plt.show()

def multi_plot_2(x_values: np.ndarray, 
               y_values: list, 
               legends: list, 
               axis_config: list, 
               tick_labels=None):
    """
    Multi-line plotter with support for:
    - x_values: common x-axis data (np.array)
    - y_values: list of np.arrays to plot
    - legends: list of legend labels for each line
    - axis_config: [xlabel, ylabel, xtick_rotation, title]
    - tick_labels: x-tick labels (strings)
    """

    plt.figure(figsize=(10, 6))

    # Plot each line with a different color and marker
    for i, y in enumerate(y_values):
        plt.plot(x_values[i], y, marker='o', label=legends[i])

    # Axes labels and title
    plt.xlabel(axis_config[0])
    plt.ylabel(axis_config[1])
    plt.title(axis_config[3])

    # X-axis tick labels
    #plt.xticks(ticks=x_values, labels=tick_labels ,rotation=axis_config[2])

    # Minor ticks and grid
    plt.gca().yaxis.set_minor_locator(ticker.AutoMinorLocator())
    plt.grid(axis="y", color="black", alpha=.5, linewidth=.5)
    plt.grid(axis="y", which='minor', color="black", alpha=.1, linewidth=.5)

    plt.legend()
    plt.tight_layout()
    #plt.savefig(axis_config[3]+".jpg")
    plt.show()

    
```

```python
def heatmap(data, row_labels, col_labels, ax=None,
            cbar_kw=None, cbarlabel="", title="", **kwargs):
    """
    Create a heatmap from a numpy array and two lists of labels.

    Parameters
    ----------
    data
        A 2D numpy array of shape (M, N).
    row_labels
        A list or array of length M with the labels for the rows.
    col_labels
        A list or array of length N with the labels for the columns.
    ax
        A `matplotlib.axes.Axes` instance to which the heatmap is plotted.  If
        not provided, use current Axes or create a new one.  Optional.
    cbar_kw
        A dictionary with arguments to `matplotlib.Figure.colorbar`.  Optional.
    cbarlabel
        The label for the colorbar.  Optional.
    **kwargs
        All other arguments are forwarded to `imshow`.
    """

    if ax is None:
        ax = plt.gca()

    if cbar_kw is None:
        cbar_kw = {}

    # Plot the heatmap
    im = ax.imshow(data, **kwargs)

    # Create colorbar
    cbar = ax.figure.colorbar(im, ax=ax, **cbar_kw)
    cbar.ax.set_ylabel(cbarlabel, rotation=-90, va="bottom")

    # Show all ticks and label them with the respective list entries.
    ax.set_xticks(range(data.shape[1]), labels=col_labels,
                  rotation=45, ha="right", rotation_mode="anchor")
    ax.set_yticks(range(data.shape[0]), labels=row_labels)

    # Turn spines off and create white grid.
    ax.spines[:].set_visible(False)

    ax.set_title(title)
    ax.set_xticks(np.arange(data.shape[1]+1)-.5, minor=True)
    ax.set_yticks(np.arange(data.shape[0]+1)-.5, minor=True)
    ax.grid(which="minor", color="w", linestyle='-', linewidth=3)
    ax.tick_params(which="minor", bottom=False, left=False)

    return im, cbar


def annotate_heatmap(im, data=None, valfmt="{x:.2f}",
                     textcolors=("black", "white"),
                     threshold=None, **textkw):
    """
    A function to annotate a heatmap.

    Parameters
    ----------
    im
        The AxesImage to be labeled.
    data
        Data used to annotate.  If None, the image's data is used.  Optional.
    valfmt
        The format of the annotations inside the heatmap.  This should either
        use the string format method, e.g. "$ {x:.2f}", or be a
        `matplotlib.ticker.Formatter`.  Optional.
    textcolors
        A pair of colors.  The first is used for values below a threshold,
        the second for those above.  Optional.
    threshold
        Value in data units according to which the colors from textcolors are
        applied.  If None (the default) uses the middle of the colormap as
        separation.  Optional.
    **kwargs
        All other arguments are forwarded to each call to `text` used to create
        the text labels.
    """

    if not isinstance(data, (list, np.ndarray)):
        data = im.get_array()

    # Normalize the threshold to the images color range.
    if threshold is not None:
        threshold = im.norm(threshold)
    else:
        threshold = im.norm(data.max())/2.

    # Set default alignment to center, but allow it to be
    # overwritten by textkw.
    kw = dict(horizontalalignment="center",
              verticalalignment="center")
    kw.update(textkw)

    # Get the formatter in case a string is supplied
    if isinstance(valfmt, str):
        valfmt = ticker.StrMethodFormatter(valfmt)

    # Loop over the data and create a `Text` for each "pixel".
    # Change the text's color depending on the data.
    texts = []
    for i in range(data.shape[0]):
        for j in range(data.shape[1]):
            kw.update(color=textcolors[int(im.norm(data[i, j]) > threshold)])
            text = im.axes.text(j, i, valfmt(data[i, j], None), **kw)
            texts.append(text)

    return texts
```
